```python
import pandas as pd
import numpy as np
```

```python
#1.
df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),columns=['A', 'B', 'C'])
df['P'] = df['A'] * df['B']
print(df)
```

```python
#2.
df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),columns=['A', 'B', 'C'])
print("DataFrame:\n", df)
print()
print("Rowwise sum:\n", df.sum(axis=1))
print()
print("Columnwise sum:\n", df.sum(axis=0))
```

```python
#3.
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)),columns=['A', 'B', 'C'])
df.loc[1, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[3, 'C'] = np.nan
print("Before filling:\n", df)
df = df.fillna(df.mean())
print("After filling:\n", df)
```

```python
#4.
df = pd.DataFrame(np.random.randint(1, 20, size=(6, 4)),columns=['A', 'B', 'C', 'D'])
df.loc[1, 'A'] = np.nan
df.loc[3, 'C'] = np.nan
df.loc[5, 'B'] = np.nan
print("Before dropping:\n", df)
df = df.dropna()
print("After dropping:\n", df)
```

```python
#5.
df = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], size=10),'Value': np.random.randint(10, 100, size=10)})
print("Original DataFrame:\n", df)
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("Aggregation:\n", result)
```

```python
#6.
df = pd.DataFrame({'Product': np.random.choice(['Laptop', 'Phone', 'Tablet'], size=10),'Category': np.random.choice(['Electronics', 'Accessories', 'Furniture'], size=10),'Sales': np.random.randint(100, 1000, size=10)})
print("Original DataFrame:\n", df)
result = df.groupby('Category')['Sales'].sum()
print("Total sales by category:\n", result)
```

```python
#7.
df1 = pd.DataFrame({'ID': [1, 2, 3, 4],'Name': ['Amit', 'Raj', 'John', 'Sam']})
df2 = pd.DataFrame({'ID': [1, 2, 3, 4],'Marks': [85, 90, 70, 99]})
result = pd.merge(df1, df2, on='ID')
print(result)
```

```python
#8.
index = pd.MultiIndex.from_tuples([('A', 1), ('A', 2), ('B', 1), ('B', 2)], names=['Category', 'Number'])
df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)
print("Multiindex Dataframe:\n", df)
print("Category A:\n", df.loc['A'])
print("Category B, Number 1:\n", df.loc[('B', 1)])
print("Slicing:\n", df.loc['A':'B'])
```

```python
#9.
index = pd.MultiIndex.from_product([['A', 'B', 'C'], ['X', 'Y']],names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(10, 100, size=6)}, index=index)
print("Original DataFrame:\n", df)
result = df.groupby(level=['Category', 'SubCategory']).sum()
print("Sum by category and Subcategory:\n", result)
```

```python
#10.
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)),columns=['A', 'B', 'C'])
print("Original DataFrame:\n", df)
result = df.apply(lambda x: x * 2)
print("\nAfter doubling:\n", result)
```

```python
#11.
df = pd.DataFrame(np.random.randint(1, 20, size=(6, 3)),columns=['A', 'B', 'C'])
print("Original DataFrame:\n", df)
df['Total'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)
print("After adding Total:\n", df)
```

```python
#12.
s = pd.Series(['hello', 'python', 'pandas', 'computer', 'data'])
print("Original Series:\n", s)
result = s.str.upper()
print("Uppercase Series:\n", result)
```

```python
#13.
s = pd.Series(['hello', 'python', 'pandas', 'matplot', 'data'])
print("Original Series:\n", s)
result = s.str[:3]

print("\nFirst three characters:\n", result)
```
